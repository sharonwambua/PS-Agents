```python
import heapq
highway_graph = {
    'Nairobi': [('Naivasha', 90), ('Thika', 45)],
    'Naivasha': [('Nakuru', 68), ('Nyahururu', 72)],
    'Thika': [('Nyeri', 105)],
    'Nyeri': [('Nyahururu', 85)],
    'Nyahururu': [('Nakuru', 60)],
    'Nakuru': [('Eldoret', 156), ('Kericho', 105)],
    'Eldoret': [('Kisumu', 120)],
    'Kericho': [('Kisumu', 75)],
    'Kisumu': []
}

def uniform_cost_search(graph, start, goal):
    frontier = []
    heapq.heappush(frontier, (0, start, [start]))
    visited = {}

    print(f"Starting search from {start} to {goal}...\n")

    step = 1
    while frontier:
        current_cost, current_town, path = heapq.heappop(frontier)

        if current_town in visited and visited[current_town] < current_cost:
            continue

        visited[current_town] = current_cost

        print(f"Step {step}: Visiting {current_town} | Total distance so far: {current_cost} km")
        print(f"        Path: {' -> '.join(path)}")
        step += 1

        if current_town == goal:
            print("\nDestination reached!")
            return path, current_cost

        for neighbor, distance in graph.get(current_town, []):
            new_cost = current_cost + distance
            new_path = path + [neighbor]
            heapq.heappush(frontier, (new_cost, neighbor, new_path))

    return None, float("inf")

start_town = 'Nairobi'
destination = 'Kisumu'

best_path, total_distance = uniform_cost_search(highway_graph, start_town, destination)

print("\n" + "="*30)
print("FINAL RESULTS")
print("="*30)
print("Best Route:", " -> ".join(best_path))
print(f"Total Cost: {total_distance} km")
```
